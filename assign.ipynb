```python
#emailsplit
emails=[
    "shinu15@link.edu.in","abc@gmail.com","prajul@iitb.ac.in","anu123@outlook.com"
]
for email in emails:
    parts=email.split("@")
    username= parts[0]
    domain= parts[1]
    print("email :",email)
    print("name :",username)
    if domain.endswith(".in"):
        print("country= india")
    elif domain.endswith(".com"):
        print("country= international")
    else:
        print("country= unknown")
    print("domain :", domain)
    print("dob:")
    if "gmail" in domain or "yahoo" in domain or "outlook" in domain :
        print("personal")
    else :
        print("official")
    print("-"*15)
```
