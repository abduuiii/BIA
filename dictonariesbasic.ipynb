```python
keys=["name","age","city"]
values=["alice",25,"new york"]
info_dict={keys[i]:values[i]for i in range(len(keys))}
print(info_dict)
new_dict={}
for i in range(0,len(keys)):
    new_dict[keys[i]]=values[i]
    print(new_dict)
    #jacquard similarities
```
