## Missing Value Function

```python
def missing_value(df, col, dtype):
    
    if dtype == "num":
        df[col] = df[col].fillna(df[col].median())
        
    elif dtype == "cat":
        df[col] = df[col].fillna(df[col].mode()[0])
        
    else:
        print("Use 'num' or 'cat'")
        
    return df
```

```python
df = missing_value(df, "bedrooms", "num")
print(df)
```

## EDA FUNCTION

```python
def eda(df):
    print("Shape of Data:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nNumerical Columns:")
    numcol = df.select_dtypes(include="number").columns.tolist()
    print(numcol)

    print("\nCategorical Columns:")
    catcol = df.select_dtypes(include="object").columns.tolist()
    print(catcol)

    print("\nNumber of Numerical Columns:", len(numcol))
    print("Number of Categorical Columns:", len(catcol))

    print("\nUnique Values in Categorical Columns:")
    for col in catcol:
        print(col, ":", df[col].nunique())

    return numcol, catcol
```

## Outlier Capping Function


```python
def cap_outliers(df, col, lower=0.05, upper=0.95):
    
    lower_limit = df[col].quantile(lower)
    upper_limit = df[col].quantile(upper)

    df[col] = df[col].clip(
        lower=lower_limit,
        upper=upper_limit
    )

    print("Column:", col)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)

    return df
```

```python
df = cap_outliers(df, "bedrooms")
```

## Feature Engineering Function

```python
def feature_engineering(df, numcol, catcol):

    print("Feature Engineering Started")

    # Numerical missing values
    for col in numcol:
        print("\nProcessing numerical column:", col)

        missing = df[col].isnull().sum()
        print("Missing values:", missing)

        df[col] = df[col].fillna(df[col].median())

        # Outlier limits
        lower_limit = df[col].quantile(0.05)
        upper_limit = df[col].quantile(0.95)

        print("Lower limit:", lower_limit)
        print("Upper limit:", upper_limit)

        # Capping
        df[col] = df[col].clip(
            lower=lower_limit,
            upper=upper_limit
        )

    # Categorical missing values
    for col in catcol:
        print("\nProcessing categorical column:", col)

        missing = df[col].isnull().sum()
        print("Missing values:", missing)

        df[col] = df[col].fillna(df[col].mode()[0])

    print("\nFeature Engineering Completed")

    return df
```
