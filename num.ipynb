```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
data={
    'A':[1,2,3,4,5],
    'B':[5,4,3,2,1],
    'C':[2,3,4,5,6],
    'D':[3,2,5,4,1]
}
df=pd.DataFrame(data)
correlation_matrix=df.corr()
print("correlation matrix:")
print(correlation_matrix)
plt.figure(figsize=(10,6))
sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm",fmt=".2f",linewidths=0.5)
plt.title("correlation matrix")
plt.show
```

```python
import numpy as np
from scipy import stats

np.random.seed(42)

# --- Sample Data ---
# Scores of two independent groups
method_A = np.random.normal(loc=75, scale=8, size=35)
method_B = np.random.normal(loc=78, scale=7, size=35)

# --- Independent Two-Sample t-test ---
t_stat, p_value = stats.ttest_ind(method_A, method_B, equal_var=False)  # Welch's t-test

print("Mean A:", np.mean(method_A))
print("Mean B:", np.mean(method_B))
print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject Null - Means are significantly different.")
else:
    print("Fail to Reject Null - No significant difference between means.")
```
