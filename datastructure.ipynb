```python
marks=[89,56,88,7,10,44,55,90]
total=0
for total in marks:
    total=sum(marks)
    average=total/len(marks)
print("Average mark:",average)
largest = max(marks)
print("Largest number:", largest)
lowest= min(marks)
print("lowest number:",lowest)
print("Numbers above average:")
for num in marks:
    if num > average:
        print(num)
```

```python
list=[100,45,65,17,88,44]
total=0
list.sort(reverse=True)
print(list)
total = sum(list)
cumulative = 0

for num in list:
    cumulative += num
    percentage = (cumulative / total) * 100
    print("Cumulative Sum =", cumulative,"  "
          "Total Percentage =", round(percentage, 2), "%")
```
